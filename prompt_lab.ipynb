## Experiment 3 — ask for a shape

Ask for a strict output format, then check whether you actually got it.

```python
answer = ask(
    "List 3 risks of committing an API key to a public repository. "
    "Return ONLY a JSON array of objects with keys 'risk' and 'severity'. ",
    label="3 format constraint",
)
```

```python
# Did it really return parseable JSON? Find out rather than assume.
import json

try:
    parsed = json.loads(answer)
    print("Parsed cleanly:", parsed)
except json.JSONDecodeError as exc:
    print("NOT valid JSON:", exc)
    print("\nWhat we actually got back:\n")
    print(repr(answer[:300]))
```
